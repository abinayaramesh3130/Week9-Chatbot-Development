
# Week 9: Healthcare Chatbot Development

## Objective
To develop a healthcare information chatbot using Ollama and the Llama 3.2 model.

## Domain
Healthcare

## Tools Used
- Python
- Ollama
- Llama 3.2
- CSV files
- Jupyter Notebook

## Features
- Answers general healthcare questions
- Provides sample questions
- Saves conversation history
- Stores chatbot responses
- Maintains activity logs

## Safety
The chatbot provides general educational information and does not replace professional medical advice.


```python

import os
import csv

project_dir = os.path.dirname(os.getcwd())

dataset_dir = os.path.join(project_dir, "Dataset")

print("Healthcare Chatbot Project")
print("Dataset folder:", dataset_dir)
print("Dataset folder exists:", os.path.exists(dataset_dir))
```

```python

questions_file = os.path.join(dataset_dir, "sample_questions.csv")

if os.path.exists(questions_file):
    with open(questions_file, "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        questions = list(reader)

    print("Sample Healthcare Questions:")
    for row in questions:
        if row:
            print("-", row[0])
else:
    print("Sample questions file not found.")
```
